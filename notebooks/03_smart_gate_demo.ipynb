# SmartGate: 1:N Biometric Access Control & Threshold Calibration

In this notebook, you will evaluate and calibrate **`SmartGate`**, a 1:N biometric verification system powered by your trained `FaceEmbeddingNet`. 

Biometric access control operates as an open-set verification system running across two distinct stages:

---

### The Two-Stage Biometric Pipeline

![training_process](assets/gate.jpeg)



#### Stage 1: Gallery Enrollment (Offline Setup)
- Authorized staff provide reference images.
- Images pass through the trained model, producing 128-D normalized embeddings $z_{\text{gallery}} \in \mathbb{R}^{128}$ on the unit hypersphere ($\|z\|_2 = 1$).
- These embeddings are stored alongside identity labels in the reference gallery database:
  $$\mathcal{G} = \{(y_i, z_i)\}_{i=1}^N$$

#### Stage 2: Real-Time Verification (At the Gate)
1. **Feature Extraction:** A live camera captures an incoming probe image, computing its embedding $z_{\text{probe}}$.
2. **1:N Nearest-Neighbor Search:** The gate computes the squared Euclidean distance to all enrolled gallery identities and selects the closest reference:
   $$d_{\text{min}} = \min_{(y_i, z_i) \in \mathcal{G}} \|z_{\text{probe}} - z_i\|_2^2$$
3. **Threshold Decision ($\tau$):**
   - **Access Granted** (Authorized Staff): If $d_{\text{min}} \le \tau$, the person is matched to identity $y_{\text{match}}$.
   - **Access Denied** (Impostor / Reject): If $d_{\text{min}} > \tau$, the person is rejected as an unknown impostor.

---

### Calibration & Parametric Threshold Sweep ($\tau$)

The decision threshold $\tau$ controls the operational security stance of the gate:
- **Too Low ($\tau \to 0$):** High security, but genuine staff are repeatedly locked out (High FRR).
- **Too High ($\tau \to 2$):** Easy entry, but impostors can slip into the facility (High FAR).

To establish an optimal operating point, you will perform a **parametric threshold sweep** across $\tau \in [0.05, 1.60]$ to measure:
- **True Acceptance Rate (TAR):** $\frac{\text{Authorized Staff Accepted}}{\text{Total Authorized Probes}}$ (higher is better)
- **False Acceptance Rate (FAR):** $\frac{\text{Impostors Accepted}}{\text{Total Impostor Probes}}$ (lower is better)
- **False Rejection Rate (FRR):** $1 - \text{TAR}$
- **Balanced Accuracy:** $\frac{\text{TAR} + (1 - \text{FAR})}{2}$
